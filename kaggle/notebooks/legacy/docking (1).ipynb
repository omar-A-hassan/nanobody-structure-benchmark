```python
from Bio.PDB import PDBIO, Select

class ChainSelect(Select):
    def __init__(self, chain_ids):
        self.chain_ids = set(chain_ids)

    def accept_chain(self, chain):
        return chain.id in self.chain_ids


io = PDBIO()
io.set_structure(structure)

# Experimental antigen = chain A
io.save(
    "8CYB_experimental_antigen.pdb",
    ChainSelect(["A"])
)

# Experimental nanobody = chain D
io.save(
    "8CYB_experimental_nanobody.pdb",
    ChainSelect(["D"])
)

print("Created:")
print("  8CYB_experimental_antigen.pdb")
print("  8CYB_experimental_nanobody.pdb")
```

```python
from Bio.PDB import PDBParser
import numpy as np

parser = PDBParser(QUIET=True)
structure = parser.get_structure("8CYB", pdb_path)

antigen = structure[0]["A"]
nanobody = structure[0]["D"]

antigen_atoms = [
    atom for residue in antigen
    if residue.id[0] == " "
    for atom in residue
]

nanobody_atoms = [
    atom for residue in nanobody
    if residue.id[0] == " "
    for atom in residue
]

min_distance = float("inf")

for a in antigen_atoms:
    for b in nanobody_atoms:
        distance = np.linalg.norm(
            a.coord - b.coord
        )

        if distance < min_distance:
            min_distance = distance

print("Minimum antigen–nanobody distance:",
      round(min_distance, 3), "Å")
```

```python
from Bio.PDB import PDBParser

def inspect_pdb(path):
    parser = PDBParser(QUIET=True)
    structure = parser.get_structure("structure", path)

    print(f"\n{path}")

    for model in structure:
        for chain in model:
            residues = [
                r for r in chain
                if r.id[0] == " "
            ]

            print(
                f"Chain {chain.id}: "
                f"{len(residues)} residues"
            )

inspect_pdb("8CYB_experimental_antigen.pdb")
inspect_pdb("8CYB_IgFold_nanobody.pdb")
```
